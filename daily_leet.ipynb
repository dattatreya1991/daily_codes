```python
def find_max_equal_array(nums):
    print(f"Array: {nums}")
    
    # Keep track of where we first saw each balance value
    balance_positions = {0: -1}  # Balance starts at 0 before we begin
    
    max_length = 0
    balance = 0
    
    print(f"\nStep by step:") 
    print(f"Position | Value | Balance | Action")
    print(f"-" * 40)
    
    for i, num in enumerate(nums):
        # Update balance
        if num == 1:
            balance += 1  # Go up for 1
        else:
            balance -= 1  # Go down for 0 
        
        if balance in balance_positions:
            # We've seen this balance before!
            old_position = balance_positions[balance]
            length = i - old_position
            max_length = max(max_length, length)
            print(f"{i:8} | {num:5} | {balance:7} | Found length {length}!")
        else:
            # First time seeing this balance
            balance_positions[balance] = i
            print(f"{i:8} | {num:5} | {balance:7} | First time seeing this balance")
    
    return max_length
```

```python
def count_subarrays_with_sum_k(nums, k):
    print(f"Array: {nums}, Target sum: {k}")
    
    # Count how many times we've seen each sum
    sum_counts = {0: 1}  # We've seen sum=0 once (before starting)
    
    current_sum = 0
    total_count = 0
    
    print(f"\nStep by step:")
    print(f"Position | Value | Current Sum | Looking for | Found | Total Count")
    print(f"-" * 70)
    
    for i, num in enumerate(nums):
        current_sum += num
        
        # What sum do we need to have seen before?
        needed_sum = current_sum - k
        
        # How many times have we seen that sum?
        if needed_sum in sum_counts:
            found_count = sum_counts[needed_sum]
            total_count += found_count
            print(f"{i:8} | {num:5} | {current_sum:11} | {needed_sum:11} | {found_count:5} | {total_count}")
        else:
            print(f"{i:8} | {num:5} | {current_sum:11} | {needed_sum:11} | {0:5} | {total_count}")
        
        # Remember this current sum for future positions
        sum_counts[current_sum] = sum_counts.get(current_sum, 0) + 1
    
    return total_count
```

```python
def max_equal_subarray(nums):
    """Simple version for interview"""
    balance_map = {0: -1}
    balance = 0
    max_len = 0
    
    for i, num in enumerate(nums):
        balance += 1 if num == 1 else -1
        
        if balance in balance_map:
            max_len = max(max_len, i - balance_map[balance])
        else:   
            balance_map[balance] = i
    
    return max_len

max_equal_subarray([0,1,0,0,1,1,0])
```

```python
def subarray_sum_k(nums, k):
    sum_count = {0: 1}
    current_sum = 0
    count = 0
    
    for num in nums:
        current_sum += num
        if (current_sum - k) in sum_count:
            count += sum_count[current_sum - k]
        sum_count[current_sum] = sum_count.get(current_sum, 0) + 1
        
    
    
    return count

# Example usage
nums = [1, 2, 3, 4, 5]
k  = 5
result = subarray_sum_k(nums, k)
print(f"\nTotal subarrays with sum {k}: {result}")
```

```python
def longest_common_prefix(strings):
    if not strings:
        return ""
    
    prefix = strings[0]  # Start with first string
    for string in strings[1:]:
        # Keep removing characters from prefix until it matches
        while prefix and not string.startswith(prefix):
            prefix = prefix[:-1]
        if not prefix:
            break
    return prefix

lists = ['flower','flower','floss','floor']
longest_common_prefix(lists)
```

```python
def merge_intervals(intervals):
    """
    Merges all overlapping intervals from a list.

    How it works:
    1. Sorts the intervals based on their start time. This is the crucial first step.
    2. Initializes a 'merged' list with the very first interval.
    3. Iterates through the rest of the sorted intervals.
    4. For each interval, it checks if it overlaps with the *last* interval in the 'merged' list.
    5. If it overlaps, it updates the end of the last merged interval. Otherwise, it adds the new interval to the list.
    """
    # Step 1: Handle edge case and sort the intervals by their start time.
    if not intervals:
        return []
    
    intervals.sort(key=lambda x: x[0]) # whats the meaning of this code ?
    print(intervals)
    
    # Step 2: Initialize the merged list with the first interval.
    merged = [intervals[0]]
    print(f'Merged is : {merged}')
    
    # Step 3: Iterate from the second interval to the end.
    for i in range(1, len(intervals)):
        current_interval = intervals[i]
        print(f'cur is: {current_interval}')
        last_merged_interval = merged[-1]
        print(f'previous is: {last_merged_interval}')
        
        # Step 4: Check for overlap.
        # An overlap occurs if the current interval's start is less than or equal to the last merged interval's end.
        if current_interval[0] <= last_merged_interval[1]:
            # Step 5: Merge by updating the end of the last interval.
            # We take the maximum of the two end points to ensure the merged interval covers both.
            last_merged_interval[1] = max(last_merged_interval[1], current_interval[1])
        else:
            # No overlap, so just add the current interval to the list.
            merged.append(current_interval)
            
    return merged


```

```python
def insert_interval(intervals, new_interval):
    """
    Inserts a new interval into a sorted list of non-overlapping intervals, merging if necessary.

    How it works:
    1. Creates a new list for the result.
    2. Iterates through the original intervals, adding all intervals that end before the new one starts.
    3. Enters a merging phase, combining the new interval with any it overlaps.
    4. Adds the final merged interval to the result.
    5. Appends any remaining intervals from the original list.
    """
    result = []
    i = 0
    n = len(intervals)

    # Phase 1: Add all intervals that end before the new interval starts.
    while i < n and intervals[i][1] < new_interval[0]:
        # print(intervals[i][1], new_interval[0])
        result.append(intervals[i])
        i += 1
    print(i, result)
    
    # Phase 2: Merge the new interval with all overlapping intervals.
    # An overlap exists as long as the current interval starts before the new one ends.
    while i < n and intervals[i][0] <= new_interval[1]:
        # Update the new_interval to be the union of itself and the current one.
        new_interval[0] = min(new_interval[0], intervals[i][0])
        new_interval[1] = max(new_interval[1], intervals[i][1])
        print(new_interval[0],intervals[i][0])
        print(f'intervals : {[new_interval[0], new_interval[1]]}')
        i += 1
    
    # After the loop, the new_interval has been fully merged. Add it to the result.
    result.append(new_interval)

    # Phase 3: Add all remaining intervals that come after the new interval.
    while i < n:
        result.append(intervals[i])
        i += 1
        
    return result


```
